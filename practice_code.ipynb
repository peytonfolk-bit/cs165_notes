```python
number_1 = 5 #int
number_2 = 2.5 #float
number_1 = number_1 + number_2 # Implicit conversion to float
# Resulting data type will be a float (7.5), this can be checked by using type()
print(type(number_1))
print(number_1)

number_3 = 3
print(float(number_3))
print(type(number_3))
# Will print 3.0
```

```python
number_3 = 3
number_3 = float(number_3)
print(number_3)
print(type(number_3))
# Will print 3.0 and actually change the integer to a float
```
