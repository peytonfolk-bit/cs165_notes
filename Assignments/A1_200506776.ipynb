```python
# Creating my variables
g = 9.8
h0 = 1.6
v0 = 14.2
t = 0.5

# Finding the height and velocity after 0.5 seconds 
h = h0 + (v0 * t) - (1/2 * g * t**2)
v = v0 - (g * t)

print(f"At {t} seconds, the height is {round(h, 2)} m.")
print(f"At {t} seconds, the velocity is {round(v, 2)} m/s.")

# Changing my time to 2 seconds!
t = 2.0

# Finding the height and velocity after 2 seconds 
h = h0 + (v0 * t) - (1/2 * g * t**2)
v = v0 - (g * t)

print(f"At {t} seconds, the height is {round(h, 2)} m.")
print(f"At {t} seconds, the velocity is {round(v, 2)} m/s.")
```
